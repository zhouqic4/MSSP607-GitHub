```python

def comma_code(items):
  #try to return an empty string result when list has no items
  #why not printable :(

  items= [str(item) for item in items]
  if len(items) == 0:
    return ''

  elif len(items) == 1:
    return items[0]

  elif len(items) == 2:
    return items[0] + ' and ' + items[1]

  else:
    return ', '.join(items[:-1]) + ', and ' + items[-1]

```

```python
spam = ['apples', 'bananas', 'tofu', 'cats']
comma_code(spam)
```
