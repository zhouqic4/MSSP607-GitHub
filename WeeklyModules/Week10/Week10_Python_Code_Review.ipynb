## 3. Show what assignment, copying, and mutation do

**Original cells 57–62 and 84–89.** Describe variables as names bound to objects. An argument binds a local parameter name to the same object. A function can mutate a shared list, and rebinding the parameter leaves the caller's name bound to its original object. Keep the original `append` example to demonstrate the side effect and show `list.copy()` when an independent outer list is needed. Use `is` for object identity and `==` for value equality.

```python
values = [1, 2, 3]
alias = values
copied_values = values.copy()

print(f"Copy has the same values before append: {copied_values == values}")
print(f"Copy is the same object: {copied_values is values}")

values.append(4)

print(f"Alias is the same object: {alias is values}")
print(f"Alias after append: {alias}")
print(f"Copy after append: {copied_values}")
```

A shallow copy creates a new outer list and shares its nested objects. This matters for examples like cell 88, which contains a list inside a list. A tuple likewise fixes which objects occupy its slots, and a mutable object inside a tuple can still change.

```python
rows = [[1, 2], [3, 4]]
copied_rows = rows.copy()
rows[0].append(5)

print(f"Separate outer lists: {copied_rows is not rows}")
print(f"Shared first row: {copied_rows[0] is rows[0]}")
print(f"Copied rows after nested append: {copied_rows}")
```

```python
import numpy as np

rng = np.random.default_rng(12345)
draws = rng.standard_normal(7)
samples_by_index = dict(enumerate(draws.tolist()))

for index, sample in samples_by_index.items():
    print(f"Sample {index}: {sample:.4f}")
```

## 7. Preserve the distinction between zero and missing

**Original cells 119–121.** Keep the explicit `is not None` check in the optional-multiplier example. Make the function runnable, give its parameters names that show their roles, and demonstrate what happens when the caller passes zero.

```python
def add_and_maybe_multiply(left, right, multiplier=None):
    '''Add two values and apply a multiplier when supplied.'''
    total = left + right
    if multiplier is not None:
        total *= multiplier
    return total


print(f"Without a multiplier: {add_and_maybe_multiply(2, 3)}")
print(f"With a zero multiplier: {add_and_maybe_multiply(2, 3, multiplier=0)}")
print(f"With a multiplier of 4: {add_and_maybe_multiply(2, 3, multiplier=4)}")
```
