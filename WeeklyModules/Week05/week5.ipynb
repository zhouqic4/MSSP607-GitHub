a. We will look at stock data from Facebook, Apple, Amazon, Netflix, and Google from different files.  We will combine them into a single file and store the dataframe for further analysis.

```python
import pandas as pd
```

b. Read each file into a dataframe

```python
fb = pd.read_csv('data/fb.csv')
aapl = pd.read_csv('data/aapl.csv')
amzn = pd.read_csv('data/amzn.csv')
nflx = pd.read_csv('data/nflx.csv')
goog = pd.read_csv('data/goog.csv')
```

```python
aapl.head()
```

```python
print('fb:', fb.shape)
print('aapl:', aapl.shape)
print('amzn:', amzn.shape)
print('nflx:', nflx.shape)
print('goog:', goog.shape)
```

c. Add a column to each dataframe, called 'ticker', indicating the ticker symbol it is for (Apple's is AAPL). This symbol is how you will look up a stock.  Each file name is the ticker symbol.

```python
fb['ticker'] = 'FB'
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
nflx['ticker'] = 'NFLX'
goog['ticker'] = 'GOOG'


nflx.head()
```

d. Append each file together into a single dataframe.

```python
faang = pd.concat([fb, aapl, amzn, nflx, goog], ignore_index=True)
faang
```

```python
print(faang.shape)
faang['ticker'].value_counts()
```

e. Save the results in a CSV file called 'faanf.csv'.

```python
faang.to_csv('data/faanf.csv', index=False)
```

f. Using 'fanng.csv', use type conversion to change the 'date' column into a 'datetime' and the 'volume' column into integers.  Then sort by 'date' and 'ticker.'

```python
faang = pd.read_csv('data/faanf.csv')
faang.dtypes
```

```python
faang['date'] = pd.to_datetime(faang['date'])
faang['volume'] = faang['volume'].astype(int)
faang.dtypes
```

```python
faang = faang.sort_values(by=['date', 'ticker'])
faang.head(10)
```

g. Find the seven rows within the highest value for volume.

```python
faang.sort_values(by='volume', ascending=False).head(7)
```

h. Use the melt() method to make the 'date' and 'ticker' key identifiers.  Melt the remaining columns so that there are not separated.

```python
melted = faang.melt(id_vars=['date', 'ticker'], value_vars=['open', 'high', 'low', 'close', 'volume'])
melted.head(10)
```
