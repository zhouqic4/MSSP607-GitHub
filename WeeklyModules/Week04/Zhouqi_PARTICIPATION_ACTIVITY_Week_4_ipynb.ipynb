```python
import os
import numpy as np
import pandas as pd
pd.options.display.max_rows = 20
```

```python
url = 'https://raw.githubusercontent.com/zhouqic4/MSSP607-GitHub/main/data/parsed.csv'

df = pd.read_csv(url)
print(df.shape)
df.columns
```

```python
#2a
results = df[(df['parsed_place'] == 'Japan') & (df['magType'] == 'mb')]
print(len(results), 'earthquakes in Japan with magType mb')
results[['parsed_place', 'mag']].head()
```

```python
#2a continued
percentile_95 = np.percentile(results['mag'], 95)
print('The 95th percentile of earthquakes in Japan is', percentile_95)
```

```python
#2b
indonesia = df[df['parsed_place'] == 'Indonesia']

indonesia['tsunami'].value_counts(normalize=True)*100
```

```python
#2b continued
pct_tsunami = indonesia['tsunami'].mean()*100
print(f'{pct_tsunami:.2f}% of Indonesia earthquakes were coupled with tsunamis')
```

```python
#2c
df[df['parsed_place'] == 'Nevada'].describe()
```

```python
#2d
ring_of_fire_places = [
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica', 'Guatemala',
    r'(?<!New )Mexico',
    'Japan', 'Philippines', 'Indonesia', 'New Zealand',
    'Antarctic',
    'Canada', 'Fiji', 'Alaska', 'Washington', 'California',
    'Russia', 'Taiwan', 'Tonga', 'Kermadec Islands'
]
df['ring_of_fire'] = df['parsed_place'].str.contains('|'.join(ring_of_fire_places), regex=True)
df[['parsed_place', 'ring_of_fire']].head()
```

```python
print(sorted(df.loc[df['ring_of_fire'], 'parsed_place'].unique()))
print('New Mexico tagged?', df.loc[df['parsed_place'] == 'New Mexico', 'ring_of_fire'].any())
```

```python
#2e
df['ring_of_fire'].value_counts().rename(index={True: 'Ring of Fire', False: 'Outside'})
```

```python
#2f
df.groupby('ring_of_fire')['tsunami'].sum()
```

```python
ring_tsunamis = df.loc[df['ring_of_fire'], 'tsunami'].sum()
print(f'Tsunamis along the Ring of Fire: {ring_tsunamis} (out of {df["tsunami"].sum()}total)')
```
