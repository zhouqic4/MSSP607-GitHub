```python
import numpy as np
import pandas as pd
pd.options.display.max_rows = 20
```

```python
url = 'https://raw.githubusercontent.com/zhouqic4/MSSP607-GitHub/main/data/Student%20Performance%20Data.csv'
students = pd.read_csv(url)
print(students.shape)
students.head()
```

```python
average = lambda x, y, z: (x + y + z) / 3
average(72, 90, 88)
```

```python
high_scorers= lambda df: df[df['total score'] >= 250]
top = high_scorers(students)
print(len(top), 'students scored 250 or more')
top.head()
```

```python
score_table = lambda df: pd.DataFrame({
    'gender': df['gender'],
    'prep course': df['test preparation course'],
    'average score': (df['math score'] + df['reading score'] + df['writing score'])/3
})
scores = score_table(students)
scores.head()
```

```python
edu_to_level = {
  'some high school': 'high school',
  'high school': 'high school',
  'some college': 'college',
  "associate's degree": 'college',
  "bachelor's degree": 'college',
  "master's degree": 'graduate'
}
```

```python
scores['parent education'] = students['parental level of education'].map(lambda x: edu_to_level[x])
scores.head()
```

```python
bins = [0, 60, 70, 80, 90, 101]
group_names = ['F', 'D', 'C', 'B', 'A']
get_grade = lambda df: pd.cut(df['average score'], bins, labels=group_names, right=False)
scores['grade'] = get_grade(scores)
scores.head()
```

```python
scores['grade'].value_counts()
```

```python
transform = lambda x: x.replace(' ', '_').upper()
scores.rename(columns=transform).head()
```

```python
prep_summary = lambda df: df.groupby('test preparation course')['total score'].mean()
prep_summary(students)
```
