## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
#sample size define with n
n=len(salaries)
n
#mean is sum of n variables/n
mean= sum(salaries)/n
print(mean)

#statistics verify
import statistics
print(statistics.mean(salaries))
```

### median

```python
#50th percentile
print(salaries)
ordered_salaries =sorted(salaries)

midpoint=len(ordered_salaries)//2
print(midpoint)
median=(ordered_salaries[midpoint-1]+ordered_salaries[midpoint])/2
print(median)

#verify
print(statistics.median(salaries))
```

### mode

```python
counts={}
for salary in salaries:
  counts[salary]=counts.get(salary,0)+1
  highest= max(counts.values())
  mode=max(counts,key=counts.get)
  print(mode, highest)

  print(statistics.mode(salaries))
```

### sample variance
Remember to use Bessel's correction.

```python
variance= sum((salary-mean)**2 for salary in salaries)/(n-1)
print(variance)
#verify
print(statistics.variance(salaries))
```

### sample standard deviation
Remember to use Bessel's correction.

```python
standardDeviation=variance**0.5
print(standardDeviation)
print(statistics.stdev(salaries))
```

### interquartile range

```python
#IQR=Q3-Q1

ordered_salaries =sorted(salaries)

position=(n-1)* 0.25
lower=int(position)
weight=position-lower
q1= ordered_salaries[lower]*(1-weight)+ordered_salaries[lower+1]*weight

position=(n-1)* 0.75
lower=int(position)
weight=position-lower
q3= ordered_salaries[lower]*(1-weight)+ordered_salaries[lower+1]*weight

IQR=q3-q1
print(q1,q3,IQR)
```

### quartile coefficent of dispersion

```python
#qcd=(q3-q1)/(q3+q1)

qcd=(q3-q1)/(q3+q1)
print(qcd)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
mean_scaled=sum(min_max_scaling)/n
cov=sum(
    (x-mean)*(y-mean_scaled) for x,y in zip(salaries,min_max_scaling)
)/(n-1)
print(cov)
#verify
print(statistics.covariance(salaries, min_max_scaling))
```

### Pearson correlation coefficient ($\rho$)

```python
standardDeviation_scaled=(sum((y-mean_scaled)**2 for y in min_max_scaling)/(n-1))**0.5
rho=cov/(standardDeviation*standardDeviation_scaled)
print(rho)
#verify
print(statistics.correlation(salaries,min_max_scaling))
```
