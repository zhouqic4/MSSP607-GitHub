a. When does the code in a function execute: when the function is defined or called?

**The code executes when the function is called. The def statement only creates the function and attaches the name to it, so nothing in the body runs at that point. Each call runs the body again from the top.**

b. How many global scopes are there in a Python program? How many local scopes?

**One global scope. It is created when the program starts and destroyed when the program ends. There's one local scope for every function call, so the count depends on how many calls are running at the moment. A function that calls another function has two local scopes alive at the same time.**

c. What happens to variables in a local scope when the function call returns?

**The local scope is destroyed and the variables in it are forgotten. the names are not stored anywhere the rest of the program can reach, so using one after the call returns raises a NameError. **

d. How can you prevent a program from crashing when it gets an error?

**I will put the code that could fail in a try clause and put the code that deals with eroor in an except caluse. when the error happens the execution jumps straight to the except clause instead of stopping the program, and then keeps going with the rest of the code.**
